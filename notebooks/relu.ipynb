```python
import sys
from pathlib import Path

# ensure parent directory is on sys.path so we can import the nn package
sys.path.insert(0, str(Path('..').resolve()))

from nn.relu import ReLU
import numpy as np
```

```python
relu = ReLU()
input = np.array([-2.0, 0.0, 3.5])
grad_output = np.array([0.1, 0.2, 0.3])

forward_output = relu.forward(input)
backward_output = relu.backward(grad_output)

print("Forward output:", forward_output)
print("Backward output:", backward_output)
```
