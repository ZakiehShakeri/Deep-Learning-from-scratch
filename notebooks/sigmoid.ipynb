```python
import sys
from pathlib import Path

# ensure parent directory is on sys.path so we can import the nn package
sys.path.insert(0, str(Path('..').resolve()))

from nn.sigmoid import Sigmoid
import numpy as np
```

```python
sigmoid = Sigmoid()
inputs = np.array([-2, 0, 3])
grad_output = np.array([1, 1, 1])
outputs = sigmoid.forward(inputs)
grad_inputs = sigmoid.backward(grad_output)
print("Inputs:", inputs)
print("Outputs:", outputs)
print("Gradients:", grad_inputs)
```
