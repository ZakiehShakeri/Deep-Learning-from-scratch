```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
# generate points roughly following y = 2x + 1 with some noise
np.random.seed(0)
X = 2 * np.random.rand(100, 1)
y = 2 * X + 1 + np.random.randn(100, 1) * 0.1
true_line = 2 * X + 1
```

```python
plt.scatter(X, y, color='blue', label='Noisy Data')
plt.plot(X, true_line, color='red', label='True Line')
plt.legend()
plt.xlabel("X")
plt.ylabel("y")
plt.title("Generated Data")
plt.show()
```

```python
w = np.random.randn(1, 1)
b = np.random.randn(1, 1) 
```

```python
# training loop
lr = 0.1
for epoch in range(100):
    y_hat = X @ w + b
    loss = np.mean((y_hat - y) ** 2)
    print("epoch", epoch, "loss:", loss)
    grad_y_hat = 2 * (y_hat - y) / np.size(y_hat)
    # grad_w = dl / dy_hat * dy_hat / dw
    grad_w = X.T @ grad_y_hat
    grad_b = np.sum(grad_y_hat)
    w = w - lr * grad_w
    b = b - lr * grad_b
    print("Final weight:", w)
print("Final bias:", b)
```

```python
y_hat = X @ w + b
# plt.scatter(X, y, color='blue', label='Noisy Data')
plt.plot(X, true_line, color='red', label='True Line')
plt.plot(X, y_hat, color='green', label='Predicted Line')
plt.legend()
plt.xlabel("X")
plt.ylabel("y")
plt.title("Generated Data")
plt.show()
```
